```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
C = 2.0             # Time constant (seconds)
dt = 0.01           # Time step size (seconds)
t_end = 10.0        # Total simulation time

# Time vector and array initializations
t = np.arange(0, t_end, dt)
N = len(t)
u = np.ones(N)       # Step input at t=0 (value = 1.0)
y = np.zeros(N)      # Output vector (starts at 0)

# Simulation loop (Euler Forward)
for k in range(1, N):
    y[k] = y[k-1] + (dt / C) * (u[k-1] - y[k-1])

# Plotting the result
plt.plot(t, u, 'r--', label='Input u(t)')
plt.plot(t, y, 'b-', label='Output y(t)')
plt.axvline(x=C, color='gray', linestyle=':', label=f't = C ({C}s)')
plt.title(r'Time Domain Simulation of $H(s) = \frac{1}{1 + 2.0s}$')
plt.xlabel('Time (seconds)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.legend()
plt.show()
```
